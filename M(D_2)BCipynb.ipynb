```python
# Predict the no. of hours the battery will last based on the battery capacity --> Linear Regression
from sklearn.linear_model import LinearRegression
battery  = [[2000],[3000],[7000],[9000]]
backup = [8,12,16,20]
model = LinearRegression()
model.fit(battery,backup)
prediction = model.predict([[8000]])
print(prediction)
```

```python
#Loan Approval Prediction (Salary is the att. to predict whether the loan should be given or not) --> decision tree (logistic also works but d.t. is best)
from sklearn.tree import DecisionTreeClassifier
salary = [[20],[30],[40],[60]]
loan_status = [0,0,1,1]
model = DecisionTreeClassifier()
model.fit(salary,loan_status)
print(model.predict([[50]]))
```
